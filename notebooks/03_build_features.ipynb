```python
import pandas as pd
import numpy as np

df = pd.read_csv("../data/USGS_RG_PFDF_Inventory_v1.csv")
df = df.replace(-9999, np.nan)

d = df[["Response", "PeakI15", "HM23", "dNBR/1000", "KF",
        "FireName", "FireState", "PreviousUSGSInventory"]].copy()
d.columns = ["Response", "I15", "T", "F", "S", "fire", "state", "source"]

d["R"] = d["I15"] * 0.25   # mm/hr over 15 minutes -> mm
d = d.dropna(subset=["Response", "R", "T", "F", "S"])
print(len(d), "rows kept out of", len(df))

d["TR"] = d["T"] * d["R"]
d["FR"] = d["F"] * d["R"]
d["SR"] = d["S"] * d["R"]
d.describe()
```

```python
print(d[["F", "S", "R"]].max())
print((d["S"] == 0).sum(), "rows with S exactly 0")
```

```python
d.to_csv("../data/model_table.csv", index=False)
```
